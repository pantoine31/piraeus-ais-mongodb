#Βήμα 8: Ενοποίηση Dynamic + Static AIS

Πρώτα κάνουμε ένα απλό join και τα ενώνουμε ως εξής:

AIS dynamic (Dec 2018) με το AIS static και μοναδικό κλειδί: vessel_id

Αυτό μας δίνει ένα ενιαίο dataset κίνησης και χαρακτηριστικών πλοίου.

```python
import pandas as pd

# 1) Dynamic (Dec 2018)
dyn = pd.read_csv(
    "unipi_ais_dynamic_dec2018.csv",
    usecols=["timestamp", "vessel_id", "lon", "lat", "heading", "speed", "course"]
)

# 2) Static
static = pd.read_csv(
    "data/ais_static/ais_static/unipi_ais_static.csv",
    usecols=["vessel_id", "country", "shiptype"]
)

# 3) Join
ais = dyn.merge(static, on="vessel_id", how="left")

print("dyn rows:", len(dyn))
print("static rows:", len(static))
print("merged rows:", len(ais))
print("missing static matches:", ais["shiptype"].isna().mean())
print(ais.head(2))
```

Μετά την ενοποίηση των δυναμικών και στατικών δεδομένων AIS με βάση το κοινό πεδίο **vessel_id**, προκύπτει ένα ενιαίο σύνολο δεδομένων που διατηρεί όλες τις καταγραφές κίνησης των πλοίων για τον Δεκέμβριο του 2018.

Ο αριθμός των εγγραφών παραμένει ίδιος πριν και μετά το join, γεγονός που επιβεβαιώνει ότι η χρήση left join δεν οδήγησε σε απώλεια δυναμικών παρατηρήσεων.

Τα στατικά δεδομένα, τα οποία περιέχουν μία εγγραφή ανά πλοίο, εμπλουτίζουν πλέον τις δυναμικές εγγραφές με πρόσθετα χαρακτηριστικά όπως η χώρα και ο τύπος πλοίου.

Παράλληλα, παρατηρείται ότι ένα μικρό ποσοστό των δυναμικών εγγραφών δεν αντιστοιχίστηκε επιτυχώς με στατικά στοιχεία, με αποτέλεσμα να εμφανίζονται κενές τιμές στα πεδία country και shiptype.

Το ποσοστό αυτό είναι περίπου 8% και κρίνεται αποδεκτό, δεδομένου ότι δεν επηρεάζει τη συνολική πληρότητα των δεδομένων κίνησης.

Συνεπώς, το σύνολο δεδομένων που δημιουργήθηκε μετά το join μπορεί να χρησιμοποιηθεί ως βάση για τον επόμενο εμπλουτισμό με μετεωρολογικές πληροφορίες μέσω χωροχρονικής συσχέτισης.

#Βήμα 9: Εμπλουτισμός του ενιαίου AIS dataset με καιρό, όχι με vessel_id, αλλά χωροχρονικά: ίδιος χρόνος (σε “ώρα”) και κοντινότερο σημείο weather στο lon/lat

```python
import pandas as pd
import geopandas as gpd

# 1) Χρόνος (AIS: ms -> datetime) / Weather: ήδη datetime
ais2 = ais.copy()
ais2["dt"] = pd.to_datetime(ais2["timestamp"], unit="ms", utc=True)
ais2["t_hour"] = ais2["dt"].dt.floor("H")

w2 = gdf_w_dec.copy()
w2["dt"] = pd.to_datetime(w2["timestamp"], utc=True)
w2["t_hour"] = w2["dt"].dt.floor("H")

# 2) GeoDataFrames σε EPSG:4326
ais_gdf = gpd.GeoDataFrame(
    ais2,
    geometry=gpd.points_from_xy(ais2["lon"], ais2["lat"]),
    crs="EPSG:4326"
)

# κρατάμε λίγες weather στήλες
weather_cols = ["t_hour", "TMP", "PRMSL", "RH", "WSPD", "APCP", "geometry"]
w_small = w2[weather_cols].copy()

# 3) Reproject σε projected CRS για σωστό nearest
ais_p = ais_gdf.to_crs(3857)
w_p   = w_small.to_crs(3857)

# 4) Join ανά ώρα + nearest
out_parts = []
hours = sorted(ais_p["t_hour"].dropna().unique())

for h in hours:
    a_h = ais_p[ais_p["t_hour"] == h]
    w_h = w_p[w_p["t_hour"] == h]
    if len(a_h) == 0 or len(w_h) == 0:
        continue

    j = gpd.sjoin_nearest(
        a_h,
        w_h,
        how="left",
        distance_col="weather_dist_m"   # σε μέτρα τώρα
    )
    out_parts.append(j)

ais_weather = pd.concat(out_parts, ignore_index=True)

# 5) Έλεγχοι + ασφαλές print (χωρίς σφάλματα)
print("AIS rows before:", len(ais_p))
print("AIS rows after join:", len(ais_weather))
print("Missing weather TMP:", ais_weather["TMP"].isna().mean())

print("Columns available:", [c for c in ["t_hour","TMP","PRMSL","RH","WSPD","APCP","weather_dist_m"] if c in ais_weather.columns])
print(ais_weather[[c for c in ["timestamp","vessel_id","lon","lat","t_hour","TMP","PRMSL","RH","WSPD","APCP","weather_dist_m"] if c in ais_weather.columns]].head(3))
```

Μετά την ενοποίηση των δυναμικών και στατικών δεδομένων AIS, πραγματοποιήθηκε ο χωροχρονικός εμπλουτισμός με μετεωρολογικές πληροφορίες για τον Δεκέμβριο του 2018.

Για τον σκοπό αυτό, τα δεδομένα AIS και τα δεδομένα καιρού ομαδοποιήθηκαν σε ωριαία χρονικά διαστήματα και στη συνέχεια, για κάθε στίγμα πλοίου, αντιστοιχίστηκε το πλησιέστερο διαθέσιμο σημείο καιρού της ίδιας ώρας.

Πριν τον υπολογισμό των αποστάσεων, τα γεωχωρικά δεδομένα προβλήθηκαν σε κατάλληλο σύστημα αναφοράς, ώστε η εύρεση του κοντινότερου σημείου να είναι γεωμετρικά ορθή.

Από τα αποτελέσματα προκύπτει ότι ο χωροχρονικός συσχετισμός πέτυχε πλήρη κάλυψη, καθώς όλες οι εγγραφές AIS που συμμετείχαν στη διαδικασία εμπλουτίστηκαν με τιμές καιρού, χωρίς απώλειες ή κενά στις βασικές μετεωρολογικές μεταβλητές.

Η απόσταση μεταξύ των θέσεων των πλοίων και των αντίστοιχων σημείων καιρού κυμαίνεται σε λογικά επίπεδα (μερικά χιλιόμετρα), γεγονός που επιβεβαιώνει την ορθότητα της μεθόδου nearest-neighbor.

Το τελικό σύνολο δεδομένων περιλαμβάνει πλέον πληροφορίες κίνησης, χαρακτηριστικά πλοίων και περιβαλλοντικές συνθήκες, και μπορεί να χρησιμοποιηθεί για περαιτέρω ανάλυση ή μοντελοποίηση.
